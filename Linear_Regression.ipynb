```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# Example data
sizes = np.array([1500, 2000, 2500, 3000, 3500]).reshape(-1, 1)
prices = np.array([245, 312, 380, 455, 510])

# Creating a linear regression model
model = LinearRegression()
model.fit(sizes, prices)

# Predictions
predicted_prices = model.predict(sizes)

# Visualization
plt.scatter(sizes, prices, color='blue', label='Actual Prices')
plt.plot(sizes, predicted_prices, color='red', label='Predicted Line')
plt.xlabel('Size (sq ft)')
plt.ylabel('Price ($1000s)')
plt.legend()
plt.show()

# Print slope (m) and intercept (b)
print(f"Slope (m): {model.coef_[0]}")
print(f"Intercept (b): {model.intercept_}")
```
